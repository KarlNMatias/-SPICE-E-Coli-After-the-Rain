```python
import geopandas as gpd
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df = pd.read_csv("Tourism_Oahu.csv") 
```

```python
df['Date'] = pd.to_datetime(df['Date'])

df['Expenditure'] = pd.to_numeric(df['Expenditure'], errors='coerce')
```

```python
df = df.dropna(subset=['Expenditure'])

# Plotting
plt.figure(figsize=(14, 6))
plt.plot(df['Date'], df['Expenditure'], marker='o')
plt.title("Monthly Visitor Expenditure – Oʻahu")
plt.xlabel("Date")
plt.ylabel("Expenditure ($ million)")
plt.grid(True)
plt.tight_layout()
plt.xticks(rotation=45)
plt.show()
```
